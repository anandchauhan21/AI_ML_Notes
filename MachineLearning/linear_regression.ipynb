```python
# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Step 2: Load dataset from GitHub
url = "https://raw.githubusercontent.com/anandchauhan21/AI_ML_Notes/refs/heads/main/MachineLearning/salary_data.csv"
df = pd.read_csv(url)

print(df.head())

# Step 3: Define variables
X = df[['Experience']]
y = df['Salary']

# Step 4: Split dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Step 5: Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Step 6: Prediction
pred = model.predict(X_test)

# Step 7: Evaluation
print("MSE:", mean_squared_error(y_test, pred))

# Step 8: Visualization
plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color='red')
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Linear Regression")
plt.show()
```

## House Price Prediction


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

url = "https://raw.githubusercontent.com/anandchauhan21/AI_ML_Notes/refs/heads/main/MachineLearning/data/House_Price_Prediction.csv"
df = pd.read_csv(url)

X = df[['Area', 'Bedrooms', 'Age']]
y = df['Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)

print("Predicted Prices:", pred)
```

## Power Consumption Prediction


```python
import pandas as pd
from sklearn.linear_model import LinearRegression

url = "https://raw.githubusercontent.com/anandchauhan21/AI_ML_Notes/refs/heads/main/MachineLearning/data/Power_Consumption_Prediction.csv"
df = pd.read_csv(url)

X = df[['Voltage', 'Current', 'Temperature']]
y = df['Power']

model = LinearRegression()
model.fit(X, y)

# Predict new value
new_data = [[255, 6.5, 38]]
pred = model.predict(new_data)

print("Predicted Power:", pred)
```
